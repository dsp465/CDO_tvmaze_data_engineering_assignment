```sql
--1) First, run the following query to optimize the table
OPTIMIZE tvmaze.silver.silver_episodes
ZORDER BY (runtime, show_id);

------------------------------------
--2) Next, run the optimized SQL query 
WITH CTE AS (
    SELECT show_id
    FROM tvmaze.silver.silver_episodes
    WHERE runtime > 40
)
SELECT
    s.genre,
    COUNT(*) AS episode_count
FROM CTE f
JOIN tvmaze.silver.silver_shows s
    ON s.show_id = f.show_id
GROUP BY s.genre
ORDER BY episode_count DESC;
```
